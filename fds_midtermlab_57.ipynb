```python
import pandas as pd

# Load dataset
df = pd.read_csv('fd39c860-f96a-4990-9a71-443ae75c178c.csv')

# View data
df.head()
df.tail()

# Dataset information
df.info()
df.describe()
df.shape

# Select columns
df['Exam_Score']
df[['Hours_Studied', 'Attendance', 'Exam_Score']]

# Select rows and columns
df.loc[0:5, ['Hours_Studied', 'Exam_Score']]
df.iloc[0:5, 0:3]

# Filter data
high_score = df[df['Exam_Score'] >= 70]

# Group data
gender_score = df.groupby('Gender')['Exam_Score'].mean()

# Aggregate data
summary = df.groupby('Gender').agg(
    Avg_Score=('Exam_Score', 'mean'),
    Avg_Attendance=('Attendance', 'mean'),
    Student_Count=('Gender', 'count')
)

# Create result column
df['Result'] = df['Exam_Score'].apply(
    lambda x: 'Pass' if x >= 40 else 'Fail'
)

# Check missing values
df.isnull().sum()

# Fill missing values
df.fillna('Unknown', inplace=True)

# Sort data
df.sort_values('Exam_Score', ascending=False)

# Top students
top_students = df.nlargest(10, 'Exam_Score')
```
