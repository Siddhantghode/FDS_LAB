```python
import pandas as pd 
import matplotlib.pyplot as plt
```

```python
# 1. Load Dataset
df = pd.read_csv('StudentPerformanceFactors.csv')

print("First 5 Rows:")
print(df.head())
```

```python
# 2. Inspect Dataset
print("\nDataset Shape:")
print(df.shape)

print("\nColumn Names:")
print(df.columns)

print("\nMissing Values:")
print(df.isnull().sum())

```

```python
# 3. Line Plot
study_performance = df.groupby('Hours_Studied')['Exam_Score'].mean()

plt.figure(figsize=(8, 5))

plt.plot(
    study_performance.index,
    study_performance.values,
    marker='o'
)

plt.title('Study Hours vs Average Exam Score')
plt.xlabel('Hours Studied')
plt.ylabel('Average Exam Score')

plt.grid(True)
plt.show()
```

```python
# 4. Bar Plot
school_performance = df.groupby('School_Type')['Exam_Score'].mean()

plt.figure(figsize=(8, 5))

plt.bar(
    school_performance.index,
    school_performance.values
)

plt.title('Average Exam Score by School Type')
plt.xlabel('School Type')
plt.ylabel('Average Exam Score')

plt.show()
```

```python
# 5. Histogram
plt.figure(figsize=(8, 5))

plt.hist(
    df['Exam_Score'],
    bins=10
)

plt.title('Distribution of Exam Scores')
plt.xlabel('Exam Score')
plt.ylabel('Frequency')

plt.show()
```

```python
# 6. Scatter Plot
plt.figure(figsize=(8, 5))

plt.scatter(
    df['Attendance'],
    df['Exam_Score']
)

plt.title('Attendance vs Exam Score')
plt.xlabel('Attendance (%)')
plt.ylabel('Exam Score')

plt.grid(True)
plt.show()
```

```python
# 7. Box Plot
male_scores = df[df['Gender'] == 'Male']['Exam_Score']
female_scores = df[df['Gender'] == 'Female']['Exam_Score']

plt.figure(figsize=(8, 5))

plt.boxplot([male_scores, female_scores])
plt.xticks([1, 2], ['Male', 'Female'])

plt.title('Exam Score Distribution by Gender')
plt.xlabel('Gender')
plt.ylabel('Exam Score')

plt.show()
```

```python
# 8. Pie Chart
internet_counts = df['Internet_Access'].value_counts()

plt.figure(figsize=(6, 6))

plt.pie(
    internet_counts.values,
    labels=internet_counts.index,
    autopct='%1.1f%%',
    startangle=90
)

plt.title('Internet Access Distribution')

plt.show()
```

```python
# 9. Heatmap
correlation = df[
    [
        'Hours_Studied',
        'Attendance',
        'Sleep_Hours',
        'Previous_Scores',
        'Tutoring_Sessions',
        'Physical_Activity',
        'Exam_Score'
    ]
].corr()

plt.figure(figsize=(9, 7))

plt.imshow(
    correlation,
    cmap='coolwarm'
)

plt.colorbar()

plt.xticks(
    range(len(correlation.columns)),
    correlation.columns,
    rotation=45
)

plt.yticks(
    range(len(correlation.columns)),
    correlation.columns
)

plt.title('Correlation Heatmap')

plt.show()
```
