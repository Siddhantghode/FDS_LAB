```python
import pandas as pd 
import numpy as np 
```

```python
#LOAD  DATASET
df = pd.read_csv("StudentPerformanceFactors.csv")

df
print(df.head())
print(df.describe())
```

```python
#SELECT DATA
exam_score = df["Exam_Score"]
previous_score = df["Previous_Scores"]

exam_score.head()
```

```python
#RANGE
range_exam = exam_score.max() - exam_score.min()
range_previous = previous_score.max() - previous_score.min()

range_exam, range_previous
```

```python
#variance
var_exam = exam_score.var()
var_previous = previous_score.var()

var_exam, var_previous
```

```python
#STANDARD DEVIATION
std_exam = exam_score.std()
std_previous = previous_score.std()

std_exam, std_previous
```

```python
#COEFFICIENT OF VARIATION
cv_exam = (std_exam / exam_score.mean()) * 100
cv_previous = (std_previous / previous_score.mean()) * 100

cv_exam, cv_previous
```

```python
#INTERPRETATION
if cv_exam < cv_previous:
    "Exam Score has lower relative variability."
else:
    "Previous Scores have lower relative variability."
```
